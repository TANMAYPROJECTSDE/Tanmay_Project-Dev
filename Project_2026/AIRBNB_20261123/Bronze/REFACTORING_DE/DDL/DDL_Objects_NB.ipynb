```python
%sql
CREATE TABLE IF NOT EXISTS tanmay_projects.airbnb_bronze.ingestion_control (
  object_id              BIGINT GENERATED ALWAYS AS IDENTITY,       
  object_name            STRING ,     
  dimensions_group       STRING ,      
  watermark_column_name  STRING,        
  watermark_updated_ts   TIMESTAMP,     
  error_message          STRING ,       
  status                 STRING  ,      
  execution_ts           TIMESTAMP     
)
USING DELTA
CLUSTER BY AUTO
TBLPROPERTIES (
  'delta.checkpointPolicy' = 'v2',
  'delta.enableDeletionVectors' = 'true',
  'delta.enableRowTracking' = 'true',
  'delta.feature.deletionVectors' = 'supported',
  'delta.feature.rowTracking' = 'supported',
  'delta.parquet.compression.codec' = 'zstd')
```
