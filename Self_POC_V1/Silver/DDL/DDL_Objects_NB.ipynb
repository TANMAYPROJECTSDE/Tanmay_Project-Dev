```python
%sql
CREATE TABLE IF NOT EXISTS tanmay_projects.airbnb_silver.calendar (
  listing_id BIGINT,
  date DATE,
  available STRING COLLATE UTF8_BINARY,
  minimum_nights INT,
  maximum_nights INT,
  Country STRING COLLATE UTF8_BINARY,
  start_ts TIMESTAMP,
  end_ts TIMESTAMP,
  Curr_flag Varchar(1))
  USING delta
  CLUSTER BY AUTO
  TBLPROPERTIES (
  'delta.checkpointPolicy' = 'v2',
  'delta.enableDeletionVectors' = 'true',
  'delta.enableRowTracking' = 'true',
  'delta.feature.deletionVectors' = 'supported',
  'delta.feature.rowTracking' = 'supported',
  'delta.parquet.compression.codec' = 'zstd')
```
